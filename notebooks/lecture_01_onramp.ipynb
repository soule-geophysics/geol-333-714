## 1. A variable is a labelled box

You put a value in, you use the label later. The label is on the left of the `=`, the value on the right.

`import numpy as np` loads a maths library and gives it the short name `np`. We use it for one thing tonight:
`np.pi`, which is a more accurate π than anything you would type by hand.

Anything after a `#` is a note for humans. Python ignores it.

```python
import numpy as np

g_accepted = 9.81   # the textbook value of gravity, in m/s^2
print(g_accepted)
print(np.pi)
```

## 2. A list holds several values

Square brackets, values separated by commas. Here are three stopwatch readings of the same pendulum.

`periods[0]` is the **first** one. Python counts from zero, which is the single most common thing to trip on.
`len(periods)` is how many there are.

```python
periods = [2.006, 2.008, 2.004]   # three measured periods, in seconds

print(periods)
print(periods[0])     # the first reading
print(len(periods))   # how many readings
```

## 3. One swing, one g

From the board: a pendulum of length `l` swinging through a small angle has period

$$T = 2\pi\sqrt{\frac{l}{g}} \qquad\text{which rearranges to}\qquad g = \frac{4\pi^2 l}{T^2}$$

So one measured pair of numbers, a length and a period, is enough to compute `g`.

In the code, `**` means "raise to a power", so `np.pi**2` is π squared. The `f` before the quote marks
lets us drop variables into text inside `{ }`, and `:.3f` means "show three digits after the decimal point".

```python
L = 1.000   # pendulum length, in meters
T = 2.006   # period of one swing, in seconds

g = 4 * np.pi**2 * L / T**2

print(f'g = {g:.3f} m/s^2')
```

About 9.81 m/s², from one length and one time.

That is the whole method. Everything for the rest of the term is doing this more carefully.

## 4. The three readings do not agree. That disagreement is data.

Look back at `periods`: 2.006, 2.008 and 2.004 seconds. Three tries at the same pendulum, three answers.

The **mean** is the single best guess the list supports. Computed two ways below, by hand and with numpy,
to show they are the same number. The library only saves typing.

```python
mean_by_hand = sum(periods) / len(periods)
mean_numpy = np.mean(periods)

print(f'by hand: {mean_by_hand:.4f} s')
print(f'numpy:   {mean_numpy:.4f} s')
```

## 5. How spread out are they?

The **standard deviation** is the typical distance of one reading from the mean. A large spread means
noisy stopwatch hands, not a broken pendulum. It tells you how much to trust any single reading.

`ddof=1` asks for the *sample* standard deviation, which is the one HW0 and every statistics package
gives you by default. Without it numpy computes a slightly different quantity.

```python
spread = np.std(periods, ddof=1)

print(f'mean:               {np.mean(periods):.4f} s')
print(f'standard deviation: {spread:.4f} s')
```

## 6. Why next week takes twenty-five readings

The average of `n` readings is about `sqrt(n)` times steadier than any single reading. That quantity has a
name, the **standard error**, and we are not deriving it tonight, only naming it.

The consequence is the part that matters: five trials at each of five lengths is twenty-five readings, and
that buys an answer about five times steadier than one stopwatch click.

```python
standard_error = spread / np.sqrt(len(periods))

print(f'mean T = {np.mean(periods):.4f} +/- {standard_error:.4f} s  (n = {len(periods)})')
```
