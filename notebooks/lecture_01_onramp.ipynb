## 2. Two lists, timed two different ways

We time the same pendulum twice over, with the same stopwatch and the same thumb.

- **`single_swing`**: five attempts at timing **one** swing.
- **`five_swings`**: five attempts at timing **five** swings, each divided by 5.

Both lists are estimates of the same thing, the period of this pendulum. Put your own numbers in below,
replacing the ones already there. A list is square brackets with values separated by commas.

`single_swing[0]` is the **first** attempt. Python counts from zero. `len()` is how many there are.

```python
# Replace these with the numbers we just timed.
single_swing = [2.06, 1.94, 2.11, 1.98, 2.05]          # five attempts at ONE swing, in seconds

t_five = [10.12, 9.97, 10.06, 10.14, 9.99]             # five attempts at FIVE swings, in seconds
five_swings = [t / 5 for t in t_five]                  # ...divided by 5, to get one period each

print('single swing: ', single_swing)
print('five swings/5:', [round(t, 3) for t in five_swings])
print('first attempt at one swing:', single_swing[0])
print('how many attempts in each list:', len(single_swing), 'and', len(five_swings))
```

## 3. One swing, one g

From the board: a pendulum of length `l` swinging through a small angle has period

$$T = 2\pi\sqrt{\frac{l}{g}} \qquad\text{which rearranges to}\qquad g = \frac{4\pi^2 l}{T^2}$$

So one measured pair of numbers, a length and a period, is enough to compute `g`.

The numbers below are the hypothetical pair from the board, kept here so this matches the worked value in
HW0. In the code, `**` means "raise to a power", so `np.pi**2` is π squared. The `f` before the quote lets
us drop variables into text inside `{ }`, and `:.3f` means "three digits after the decimal point".

```python
L = 1.000   # pendulum length, in meters
T = 2.006   # period of one swing, in seconds

g = 4 * np.pi**2 * L / T**2

print(f'g = {g:.3f} m/s^2')
```

About 9.81 m/s², from one length and one time.

That is the whole method. The rest of the term applies it with more care.

## 4. The mean of each list

The **mean** is the single best guess a list supports. Computed below for both lists, by hand and with
numpy, to show the two give the same number. The library only saves typing.

The two means should be close to each other. That is expected: both lists are measuring the same pendulum.

```python
mean_single = sum(single_swing) / len(single_swing)   # by hand
mean_five = np.mean(five_swings)                      # with numpy

print(f'mean of the single-swing attempts: {mean_single:.4f} s')
print(f'mean of the five-swing attempts:   {mean_five:.4f} s')
```

## 5. The standard deviation of each list

The **standard deviation** is the typical distance of one reading from the mean of its list. It says how
much to trust any single attempt.

`ddof=1` asks for the *sample* standard deviation, the one HW0 and every statistics package gives you by
default.

Before you run it, predict which list has the smaller standard deviation, and by roughly how much.

```python
spread_single = np.std(single_swing, ddof=1)
spread_five = np.std(five_swings, ddof=1)

print(f'spread, timing one swing:   {spread_single:.4f} s')
print(f'spread, timing five swings: {spread_five:.4f} s')
print()
print(f'the single-swing list is {spread_single / spread_five:.1f} times wider')
```

Your thumb makes the same error either way: roughly a tenth of a second, starting and stopping.

Timing one swing, that whole error lands on one period. Timing five and dividing by five, the same error is
shared across five periods, so it lands on each one **five times smaller**.

That is why the data you collect for HW0 times **ten** swings and divides by ten.

## 6. Why the collection takes twenty-five readings

Repeating a measurement helps too, and it helps in a different way. The average of `n` readings is about
`sqrt(n)` times steadier than any single reading. That quantity is the **standard error**.

Five trials at each of five lengths is twenty-five readings, and the resulting mean is about five times
steadier than a single reading.

```python
standard_error = spread_five / np.sqrt(len(five_swings))

print(f'mean T = {np.mean(five_swings):.4f} +/- {standard_error:.4f} s  (n = {len(five_swings)})')
```
